```python
# Importación de librerías para manipulación de datos y visualización
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar los datos
# Se omiten las primeras 13 líneas de metadatos y se especifica que "Nulo" equivale a NaN
archivo = "25001 (1).txt"
nombres_columnas = ['FECHA', 'PRECIP_MM', 'EVAP_MM', 'TMAX_C', 'TMIN_C']

df = pd.read_csv(
    archivo, 
    sep=r'\s+',          # Separador por uno o más espacios
    skiprows=14,         # Omite el encabezado y metadatos
    names=nombres_columnas, 
    na_values=['Nulo'],  # Convierte el string 'Nulo' en NaN (nulo numérico)
    parse_dates=['FECHA'], 
    dayfirst=True        # El formato original es DD/MM/YYYY
    encoding='latin1'  # Agrega este parámetro
)

# 2. Contar los registros faltantes por fecha (antes de rellenar)
# Evaluamos cuántos datos nulos hay por fila (por fecha)
faltantes_por_fecha = df.set_index('FECHA').isnull().sum(axis=1)
fechas_con_faltantes = faltantes_por_fecha[faltantes_por_fecha > 0]

print("--- Registros faltantes por fecha (Muestra) ---")
print(fechas_con_faltantes.head(10)) 
print("\nTotal de fechas con al menos un dato nulo:", len(fechas_con_faltantes))

# 3. Eliminar nulos y rellenar con la media de su columna
# Identificamos las columnas estrictamente numéricas
cols_numericas = ['PRECIP_MM', 'EVAP_MM', 'TMAX_C', 'TMIN_C']

# Convertimos a float para evitar conflictos de tipo de dato
df[cols_numericas] = df[cols_numericas].apply(pd.to_numeric, errors='coerce')

# Rellenamos los valores NaN con el promedio (mean) calculado por cada columna
df[cols_numericas] = df[cols_numericas].fillna(df[cols_numericas].mean())

# 4. Exploración de los datos limpios
print("\n--- Primeros registros (head) ---")
display(df.head())

print("\n--- Dimensiones del dataset (shape) ---")
print(df.shape)

print("\n--- Estadísticas descriptivas (describe) ---")
display(df.describe())

# 5. Visualizaciones
# Configurar estilo global para las gráficas
sns.set_theme(style="whitegrid")

# Generar histogramas de las variables numéricas
df[cols_numericas].hist(figsize=(10, 8), bins=20, color='teal', edgecolor='black')
plt.suptitle("Distribución de variables climáticas", fontsize=16)
plt.tight_layout()
plt.show()

# Mapa de calor (Heatmap) de correlaciones
plt.figure(figsize=(8, 6))
matriz_correlacion = df[cols_numericas].corr()
sns.heatmap(matriz_correlacion, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Mapa de calor de correlaciones", fontsize=14)
plt.show()
```
