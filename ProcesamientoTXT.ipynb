```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carga de datos convirtiendo 'Nulo' a NaN
df = pd.read_csv('25001 (1).txt', sep=',', na_values='Nulo', encoding='latin1')
# Si la columna de fecha existe (usualmente se llama 'Fecha'), la convertimos a formato datetime
# df['Fecha'] = pd.to_datetime(df['Fecha'])
```

```python
# Muestra filas y columnas
print("Shape:", df.shape)

# Muestra los tipos de datos de cada columna y cuántos valores no nulos hay (muy útil)
df.info()
```

```python
# Muestra los primeros 5 y los últimos 5 registros
display(df.head())
display(df.tail())

# Cuenta cuántos datos faltantes (Nulos) hay por cada columna
print("\nValores nulos por columna:")
print(df.isnull().sum())
```

```python
# Genera estadísticas descriptivas: conteo, media, desviación estándar, mínimos, máximos y cuartiles
display(df.describe())
```

```python
# Histogramas de todas las variables numéricas
df.hist(figsize=(10, 8), bins=20)
plt.show()
```

```python
# Los diagramas de caja son excelentes para ver valores atípicos (outliers) en las temperaturas o lluvias
# Seleccionamos solo las columnas numéricas
numericas = df.select_dtypes(include=['float64', 'int64'])

sns.boxplot(data=numericas)
plt.show()
```

```python
# Compara dos variables para ver si hay relación (ej. Temperatura Máxima vs Mínima)
# NOTA: Cambia 'Tmax' y 'Tmin' por los nombres exactos de tus columnas
df.plot.scatter(x='Tmax', y='Tmin', alpha=0.5)
plt.show()
```

```python
# Ideal para datos climatológicos para ver el comportamiento a lo largo del tiempo
# NOTA: Cambia 'Precip' por el nombre exacto de la columna de precipitación
df.plot.line(y='Precip', figsize=(12, 4), title="Comportamiento de la Precipitación")
plt.show()
```

```python
# Mide qué tan relacionadas están las variables numéricas entre sí
plt.figure(figsize=(6, 4))
sns.heatmap(numericas.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.show()
```
